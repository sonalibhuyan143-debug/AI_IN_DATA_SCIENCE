## 1. MERGE(): Column based

```python
import pandas as pd
df1 = pd.DataFrame({'A':[2,3,4,5], 'B': ['A', 'B', 'C', 'D']})
df2 = pd.DataFrame({'A':[1,2,3,4], 'C': ['A1', 'B1', 'C1', 'D1']})
pd.merge(df1, df2)
```

## JOIN(): Index based

```python
import pandas as pd
df1 = pd.DataFrame({"A": [1,2,3,4,5], "B":['sonali','kanishka','sadaf', 'pragya', 'umang']})
df2 = pd.DataFrame({"A":['60','70','80','90','100']})
df1.join(df2)                                                       #it will give error as column names are same
```

```python
import pandas as pd
df1 = pd.DataFrame({"A": [1,2,3,4,5], "B":['sonali','kanishka','sadaf', 'pragya', 'umang']})
df2 = pd.DataFrame({"D":['60','70','80','90','100']})
df1.join(df2)
```

```python
import pandas as pd
df1 = pd.DataFrame({"A": [1,2,3,4,5], "B":['sonali','kanishka','sadaf', 'pragya', 'umang']}, index = ['A','B','C','D','E'])
df2 = pd.DataFrame({"C": [1,2,3,4,5], "D": ['60', '70', '80', '90','100']})
df1.join(df2)
```

```python
import pandas as pd
df1 = pd.DataFrame({"A": [1,2,3,4,5], "B":['sonali','kanishka','sadaf', 'pragya', 'umang']}, index = ['1','2','3','4','5'])
df2 = pd.DataFrame({"C": [1,2,3,4,5], "D": ['60', '70', '80', '90','100']})
df1.join(df2)
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5,6], 'B':['A','B','C','D','E']},index = ['A','B','C','D','E'])
df2 = pd.DataFrame({'D': [2,3,], 'C': ['A','C']},index= ['A','B'])
df1.join(df2)
df2.join(df1)
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5,6], 'B':['A','B','C','D','E']},index = ['A','B','C','D','E'])
df2 = pd.DataFrame({'D': [2,3,], 'C': ['A','C']},index= ['A','B'])
df1.join(df2,how = "left")
df2.join(df1,how = "left")
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5,6], 'B':['A','B','C','D','E']},index = ['A','B','C','D','E'])
df2 = pd.DataFrame({'D': [2,3,], 'C': ['A','C']},index= ['A','B'])
df1.join(df2,how = "inner")
df2.join(df1,how = "inner")
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5,6], 'B':['A','B','C','D','E']},index = ['A','B','C','D','E'])
df2 = pd.DataFrame({'D': [2,3,], 'C': ['A','C']},index= ['A','B'])
df1.join(df2,how = "outer")
df2.join(df1,how = "outer")
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5,6], 'B':['A','B','C','D','E']},index = ['A','B','C','D','E'])
df2 = pd.DataFrame({'C': [2,3,], 'D': ['A','C']},index= ['A','B'])
df1.join(df2,how = "right")
df2.join(df1,how = "right")
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5,6], 'B':['A','B','C','D','E']},index = ['A','B','C','D','E'])
df2 = pd.DataFrame({'A': [2,3,], 'D': ['A','C']},index= ['A','B'])
df1.join(df2,how = "right",lsuffix="_left")                                #using lsuffix to remove overlapping columns from left DF
df2.join(df1,how = "right",lsuffix="_left")
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5,6], 'B':['A','B','C','D','E']},index = ['A','B','C','D','E'])
df2 = pd.DataFrame({'A': [2,3,], 'D': ['A','C']},index= ['A','B'])
df1.join(df2,how = "right",rsuffix="_right")                               #using rsuffix to rename overlapping columns from right DF
df2.join(df1,how = "right",rsuffix="_right")
```

```python
import pandas as pd
df1 = pd.DataFrame({'A': [2,3,4,5],'B':['A','B','F','D']})
df2 = pd.DataFrame({'C': [2,3,4,5],'D':['A','B','F','D']})
df3 = pd.DataFrame({'E': [2,3,4,5],'F':['A','B','F','D']})
df1.join([df2,df3])

```

Perform validate operations on datasets(checking whether data is correct, valid and follows required rules)

1. One-to-one

```python
import pandas as pd
df1 = pd.DataFrame({'ID': [1,2,3,4],'name':['sonali','kanishka','sadaf','pragya']})
df2 = pd.DataFrame({'ID': [1,2,3,4],'marks':['10','20','30','40']})
result= pd.merge(df1, df2, on = 'ID', validate= 'one_to_one')
print(result)
```

2. One-to-many

```python
import pandas as pd
df1 = pd.DataFrame({'ID': [1,2,3,4],'name':['sonali','kanishka', 'sadaf','pragya']})
df2 = pd.DataFrame({'ID': [1,1,3,4],'subject':['english','hindi','science','SST']})
result= pd.merge(df1, df2, on = 'ID', validate= 'one_to_many')
print(result)
```

3. Many-to-one

```python
import pandas as pd
df1 = pd.DataFrame({'subject': ['english', 'maths', 'science', 'SST'],'marks':[10,30,50,70]})
df2 = pd.DataFrame({'subject': ['english', 'maths', 'science'],'teacher':['A','B','C']})
result= pd.merge(df1, df2, on = 'subject', validate= 'many_to_one')
print(result)
```

4. Many-to-many

```python
import pandas as pd
df1 = pd.DataFrame({'subject': ['maths', 'maths', 'science', 'SST'],'marks':[10,30,30,70]})
df2 = pd.DataFrame({'subject': ['english', 'maths', 'science'],'teacher':['A','C','C']})
result= pd.merge(df1, df2, on = 'subject', validate= 'many_to_many')
print(result)
```
