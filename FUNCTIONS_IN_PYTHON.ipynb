```python
#To print sum of two numbers
def Sum(a,b):
  sum = a + b
  return sum
print(Sum(2,3))
```

```python
def juice(fruit):
  return fruit + "juice"
juice("apple")
```

```python
def calculator(a,b):
  sum = a+b
  sub = a-b
  mul = a*b
  div = a/b
  coeff = a**b
  return sum , sub , mul , div , coeff
calculator(2,4)

```

```python
def avg(a,b,c):
  sum = a + b + c
  avg = sum/3
  return avg
avg(2,3,4)
```

```python
def avg(a,b,c=5):
  sum = a+b+c
  avg = sum/3
  return avg

avg(2,3)

```

```python
def avg(a=4,b,c):
  sum = a+b+c
  avg = sum/3
  return avg
avg(3,4)
```

```python
#factorial of a number
n = int(input("enter a number:"))    # n = global variable
def factt(n):
  fact = 1                           # fact = local variable
  for i in range(1,n+1):
    fact = fact*i
  return fact
factt(n)
```

```python
# Lambda function
square = lambda X : X*X
print(square(5))
```

```python
square = lambda X : X + X
print(square(5))
```

```python
square = lambda X,Y : X*Y
print(square(6,7))
```
