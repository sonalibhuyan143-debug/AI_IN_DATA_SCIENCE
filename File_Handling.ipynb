W+(WRITE AND READ)

```python
f = open("new_file.txt", "w+")
f.write("my age is 23")
f.write("\nsecond line")
f.seek(0)
data = f.read()
print(data)
f.close()
```

```python
f = open("sonali file.txt", "w+")     #w+ opens existing file and deletes all old contents
f.write("new data")
f.seek(0)
print(f.read())
f.close()
```

```python
#Make a file and read and check if python keyword is there or not.
f = open("sample.txt", "w")
f.write("python is a programming language.")
f.close()
f = open("sample.txt", "r")
data = f.read()
if "python" in data:
  print("python keyword is present")
else:
  print("python keyword is not present")
f.close()
```
