# **Data manipulation with Pandas**

# 1. Series

```python
import pandas as pd
marks = [2,4,6,8,10]
DS = pd.Series(marks)
print(DS)
```

```python
#customized index
import pandas as pd
marks = [2,4,6,8,10]
DS = pd.Series(marks)
print(type(DS))
print(DS)

```

```python
import pandas as pd
marks = [2,4,6,8,10]
DS = pd.Series(marks, index = ['a','b','c','d','e'])
print(type(DS))
print(DS)
```

```python
import pandas as pd
marks = [2,4,6,8,10]
DS = pd.Series(marks, index = ['a','b','c','d'])
print(type(DS))
print(DS)
```

```python
import pandas as pd
marks = [2,4,6,8,10]
DS = pd.Series(marks, index = ['a','b','c','d','e','f'])
print(type(DS))
print(DS)
```

```python
import pandas as pd
marks = [2,4,6,8,10]
DS = pd.Series(marks, index = ['a','b','c','d','e'],dtype = "float")
print(type(DS))
print(DS)
```

```python
import pandas as pd
marks = [2,4,6,8,10]
DS = pd.Series(marks,index=['a','b','c','d','e'],dtype = "float", name= "marksheet")
print(DS)
```

# 1.1) Common operation

```python
#sum
print(DS.sum())
print("sum of series:", DS.sum())
```

```python
#max
print(DS.max())
print("max no in series:", DS.max())
```

```python
#mean
print(DS.mean())
print("mean of series:", DS.mean())
```

```python
#min
print(DS.min())
print("min of series:", DS.min())
```

1.2) Indexing and slicing


```python
print("accessc element by int", DS["a"])
```

```python
variable = "a"
print("access element by variable", DS[variable])
```

```python
print("slicing in series", DS[0:3])
```
