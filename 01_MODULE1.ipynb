```python
# Write a program to create a simple calculator using if-elif-else.
num1 = float(input("enter first number:"))
num2 = float(input("enter second number:"))
op = input("enter operator (+, -, *, /):")
if op == '+':
  print("result =", num1 + num2)
elif op == '-':
  print("result =", num1 - num2)
elif op == '*':
  print("result =", num1 * num2)
elif op == '/':
  print("result =", num1 / num2)
else:
  print("division by zero is not possible")
  print("invalid operator")
```

# DICTIONARY AND SET

```python
# Write a program to create a dictionary from two lists: one of keys and one of values.
keys = ["name" , "age" , "city"]
values = ["sonali" , 23 , "delhi"]
dictionary = dict(zip(keys, values))
print("Dictionary =", dictionary)

```

```python
# Merge two dictionaries into one.
dictionary1 = {'name': 'sonali', 'age': 23, 'city': 'delhi'}
dictionary2 = {'course': 'B.tech'}
merge = dictionary1 | dictionary2
print("merged dictionary =", merge)

```

```python
# Write a program to sort a dictionary by its values.
dictionary = {'A': 15, 'B': 25, 'C': 40, 'D': 50, 'E': 60}
sorted_values = sorted(dictionary.values())
print("sorted values:", sorted_values)
```

```python
# Given two sets check if one set is subset of another.
set1 = {1,2,3}
set2 = {1,2,3,4,5}
if set1.issubset(set2):
  print("set1 is subset of set2")
else:
  print("set1 is not a subset of set2")
```

```python
#  Write a program to check whether two lists have atleast one common element using sets.
list1 = [1,2,4,5,6]
list2 = [6,7,8,9,10]
A = set(list1)
B = set(list2)
if A.intersection(B):
  print("the lists have atleast one common element")
else:
  print("the lists have no common element")
```

```python
# Write a program to count even and odd numbers in a list.
numbers = [2,3,7,8,5,1]
even = 0
odd = 0
for i in numbers:
  if i % 2 == 0:
    even += 1
  else:
    odd += 1
    print("number of even elements:", even)
    print("number of odd elements:", odd)
```
