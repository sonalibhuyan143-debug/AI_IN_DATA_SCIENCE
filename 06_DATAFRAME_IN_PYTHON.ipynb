### *Creating a data frame with list*

```python
import pandas as pd
Studentsinfo = [['A','B','C'], [20,40,60]]
DF = pd.DataFrame(Studentsinfo)
print(DF)
```

```python
import pandas as pd
Studentsinfo = [['A','B','C'], [20,40,60]]
DF = pd.DataFrame(Studentsinfo, index = ['name','marks'])
DF
```

```python
import pandas as pd
Sr = {
    "name": pd.Series(['A','B','C','D', 'E']),
    "roll_no": pd.Series([10,11,12,13,14,15]),
}
DF = pd.DataFrame(Sr)
print(DF)
```

 Arithmetic operations in data frame

```python
import pandas as pd
var = pd.DataFrame({'A': [1,2,3,4], 'B': [5,6,7,8]})
print(var)
```

```python
import pandas as pd
var = pd.DataFrame({'A': [1,2,3,4], 'B': [5,6,7,8]})
var = var["A"] + var["B"]
var
```

 Filtering in data frame

```python
import pandas as pd
var = pd.DataFrame({'A': [1,2,3,4], 'B': [5,6,7,8]})
var = var["A"] + var["B"]
var["python"] = var > 10
var
```

Handling missing values*

```python
import numpy as np
info = {
    "name": ['A','B','C','D','np.nan'],
    "roll_no": [10,20,30,40,np.nan],
    "marks": [20,30,40,np.nan,50],
     "city": ['a','b','c','d','e'],
}
DF = pd.DataFrame(info)
DF
```

```python
DF.info()
```

```python
DF.isnull()
```

```python
DF.isnull().sum()
```

```python
print(DF.isnull().sum().sum())
```

```python
#Remove null values
DF.dropna()
```

```python
DF.dropna(subset= ["marks"])
```

```python
DF["marks"]= DF["marks"].fillna(0)
DF
```

```python
DF["marks"]= DF["marks"].fillna(DF["marks"].mean())
DF
```

```python
DF["marks"] = DF["marks"].fillna(DF["marks"].median())
DF
```
